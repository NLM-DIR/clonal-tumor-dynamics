# Plot experimental mouse data

This file plots the experimental growth data of all mice, generating Figures 2a and 4a.

## Prep

Import needed packages

```python
%matplotlib widget
from matplotlib import pyplot as plt
import pandas as pd
import numpy as np
import os
```

Close any figures generated from previous runs

```python
plt.close("all")
```

## Define data and user-input parameters

Define the data file and save files

```python
data_file = "data/growth_data_df.csv"

# If you don't want to save the figures, set to None
nude_save_file = "figures/nude_growth.svg"
b6_save_file = "figures/b6_growth.svg"
```

Define the groups and the titles to use for plots

```python
groups_nude = ["Grp. B1 nude (100% C1)", "Grp. B2 nude (80% C1; 20% C11)", "Grp. B3 nude (50% C1; 50% C11)", "Grp. B4 nude (20% C1; 80% C11)", "Grp. B5 nude (100% C11)"]
groups_b6 = ["Grp. A1 B6 (100% C1)", "Grp. A2 B6 (80% C1; 20% C11)", "Grp. A3 B6 (50% C1; 50% C11)", "Grp. A4 B6 (20% C1; 80% C11)", "Grp. A5 B6 (100% C11)"]
titles = ["100% Proliferative", "80% Proliferative\n20% Invasive", "50% Proliferative\n50% Invasive", "20% Proliferative\n80% Invasive", "100% Invasive"]
```

Load the data

```python
growth_df = pd.read_csv(data_file)
```

Function to plot the nude mouse data. Shows the plot in the notebook and saves if nude_save_file is set.

```python
def plot_nude_data():
    # Define the maximum value of the y-axis to be 100 mm^3 larger than the maximum value across all nude groups
    ylim = growth_df[growth_df["group"].isin(groups_nude)]["size"].max()+100
    # Change figure size at the last part of this line if needed. First value is width, second value is height.
    fig, axes = plt.subplots(nrows=1, ncols=5, sharex=True, sharey=True, figsize=(10, 3))
    # Loop through the nude groups
    for i in range(len(groups_nude)):
        # Extract the data for that group
        group_df = growth_df[growth_df["group"] == groups_nude[i]]
        # For each mouse in that group, plot its growth on the group-specific plot
        for mid in group_df["id"].unique():
            axes[i].plot(group_df[group_df["id"]==mid]["day"], group_df[group_df["id"]==mid]["size"], color="black", linewidth=1)#, alpha=0.3)
        # Annotate the group-specific plot
        axes[i].set(title=titles[i], ylim=[-1, ylim], xlabel="Day", ylabel="Size (mm^3)")
    # Annotate the whole plot
    fig.suptitle("Immunocomprimised mice")
    plt.tight_layout()
    # Save if file is provided
    if nude_save_file:
        plt.savefig(nude_save_file)
    # Show the figure
    plt.show()
```

Function to plot the B6 mouse data. Shows the plot in the notebook and saves if b6_save_file is set.

```python
def plot_b6_data():
    # Define the maximum value of the y-axis to be 100 mm^3 larger than the maximum value across all B6 groups
    ylim = growth_df[growth_df["group"].isin(groups_b6)]["size"].max()+100
    
    # Change figure size at the last part of this line if needed. First value is width, second value is height.
    fig, axes = plt.subplots(nrows=1, ncols=5, sharex=True, sharey=True, figsize=(10, 3))
    
    # Loop through the nude groups
    for i in range(len(groups_b6)):
        # Extract the data for that group
        group_df = growth_df[growth_df["group"] == groups_b6[i]]
        # For each mouse in that group, plot its growth on the group-specific plot
        for mid in group_df["id"].unique():
            axes[i].plot(group_df[group_df["id"]==mid]["day"], group_df[group_df["id"]==mid]["size"], color="black", linewidth=1)
        # Annotate the group-specific plot
        axes[i].set(title=titles[i], ylim=[-1, ylim],  xlabel="Day", ylabel="Size (mm^3)")
    
    # Annotate the whole plot
    fig.suptitle("Immunocompetent mice")
    plt.tight_layout()
    
    # Save if file is provided
    if b6_save_file:
        plt.savefig(b6_save_file)

    # Show the figure
    plt.show()
```

Plot the nude data

```python
plot_nude_data()
```

Plot the B6 data

```python
plot_b6_data()
```
